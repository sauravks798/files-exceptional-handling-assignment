1. What is the difference between interpreted and compiled languages?
  - Compiled Languages
      * A compiler translates the entire source code into machine code before the program runs.
      * You get a separate executable file (like .exe) that can be run independently of the source code.
      * Example compiled languages: C, C++, Rust, Go.

  - Interpreted Languages
     * An interpreter reads and executes code line by line at runtime.

     * There's no separate executable; the source code is needed every time the program runs.

     * Python, Ruby, and JavaScript are typically interpreted.

9. What is the difference between multithreading and multiprocessing?
  - Multithreading
      * Uses multiple threads inside a single process.
      * The threads share memory and resources.
      * Great for tasks that spend time waiting (like downloading, reading files).

  - Multiprocessing
      * Uses multiple processes, each with its own memory and CPU core.
      * They don’t share memory (they communicate via pipes/queues).
      * Great for CPU-heavy tasks (math, data processing, crunching numbers).

11. What is memory management in Python?
  - Memory management in Python is the process of handling the allocation and deallocation of memory during program execution. Python uses a built-in system for efficient memory management.

  - Python has a garbage collector that automatically removes unused objects from memory to free up space. It uses reference counting and a cyclic garbage collector to identify and clean up objects no longer in use.

  - Every object in Python has a reference count, which tracks how many references point to that object. When the reference count drops to zero, the memory occupied by the object is automatically reclaimed.

```python
# 12. What are the basic steps involved in exception handling in Python?
''' Exception handling in Python is used to manage errors gracefully during program execution.
It allows the program to continue running or fail with a meaningful message instead of crashing unexpectedly.'''

# The basic steps are:
# try block
# except block
# else block
# finally block

try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("You can't divide by zero.")
except ValueError:
    print("Invalid input! Please enter a number.")
else:
    print("Result:", result)
finally:
    print("Execution complete.")
```

13. Why is memory management important in Python?
  - Efficient Resource Usage: It ensures that memory is used efficiently, preventing wastage and allowing programs to run smoothly, especially for large applications or data-heavy tasks.

  - Automatic Garbage Collection: Python automatically frees memory that's no longer in use. This helps avoid memory leaks, which can slow down or crash programs over time.

  - Improved Performance: Proper memory handling leads to faster program execution, as memory is allocated and deallocated in an optimized way.

  - Prevents Crashes and Freezing: Without good memory management, a program might consume more memory than available, leading to crashes or system instability.

  - Developer Convenience: Python abstracts away much of the complexity of memory handling. This allows developers to focus on writing code rather than managing memory manually.

18. What is the difference between os.fork() and multiprocessing in Python?
  - os.fork()
      * Low-level system call available only on Unix/Linux systems.
      * It creates a child process by duplicating the current process.
      * The child process is an exact copy of the parent (including memory space).
      * After forking, both parent and child continue running independently.
      * You have to manually manage communication between processes.

  - multiprocessing Module
      * High-level module available in both Unix and Windows.
      * Provides a clean, Pythonic interface for creating and managing multiple processes.
      * Supports process communication, synchronization, and shared memory easily.
      * Each process has its own memory space.

```python
# 4. Write a Python script that reads from one file and writes its content to another file.
def copy_file(source_path, destination_path):
    try:
        with open(source_path, 'r') as source_file:
            content = source_file.read()

        with open(destination_path, 'w') as destination_file:
            destination_file.write(content)

        print(f"Contents copied from '{source_path}' to '{destination_path}' successfully.")
    except FileNotFoundError:
        print(f"Error: The file '{source_path}' does not exist.")
    except IOError as e:
        print(f"An I/O error occurred: {e}")

# Example usage
source = 'file.txt'
destination = 'file2.txt'
copy_file(source, destination)
```

```python
# 6. Write a Python program that logs an error message to a log file when a division by zero exception occurs.
import logging

logging.basicConfig(
    filename='error.log',
    level=logging.ERROR,
    format='%(asctime)s - %(levelname)s - %(message)s'
)

def divide_numbers(a, b):
    try:
        result = a / b
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")

# Example usage
divide_numbers(10, 0)
```

```python
# 21. Write a Python program that reads a file and prints the number of occurrences of a specific word.
def count_word_occurrences(file_path, word):
    try:
        with open(file_path, 'r') as file:
            content = file.read()
            word_count = content.lower().count
            print(f"The word '{word}' appears {word_count} times in the file.")
    except FileNotFoundError:
        print(f"Error: The file '{file_path}' does not exist.")
    except IOError as e:
        print(f"An I/O error occurred: {e}")

# Example usage
file_path = 'file.txt'
word = 'Hello'
count_word_occurrences(file_path, word)
```

```python
# 23. Write a Python program that writes to a log file when an error occurs during file handling.
import logging

logging.basicConfig(
    filename='error.log',
    level=logging.ERROR,
    format='%(asctime)s - %(levelname)s - %(message)s'
)

def read_file(filename):
    try:
        with open(filename, 'r') as file:
            content = file.read()
            print("File content:\n", content)
    except FileNotFoundError:
        logging.error(f"File '{filename}' not found.")
        print(f"Error: File '{filename}' not found.")
    except IOError as e:
        logging.error(f"I/O error while reading '{filename}': {e}")
        print(f"Error: Could not read file '{filename}'.")
    except Exception as e:
        logging.error(f"Unexpected error while reading '{filename}': {e}")
        print("An unexpected error occurred.")

# Example usage
filename = input("Enter the filename to read: ")
read_file(filename)
```
